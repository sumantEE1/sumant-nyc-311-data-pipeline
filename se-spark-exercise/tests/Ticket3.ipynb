```python
start_time = time.time()
spark.read.csv("../data/archive/311-service-requests-from-2010-to-present-10.csv",header=True, inferSchema=True)
end_time = time.time()
```

```python
execution_time = end_time - start_time
print(f"Time taken to read CSV with header and inferred schema: {execution_time:.4f} seconds")
```

```python
service_requests = spark.read.csv("../data/archive/311-service-requests-from-2010-to-present.csv", header=True, inferSchema=True)
```

```python
from pyspark.sql.functions import col
sel_col_ser_req = service_requests.select(col("Complaint Type"), col("Borough"), col("Agency"), col("Created Date"), col("Closed Date"), col("Status"), col("Descriptor"))
```

```python
filtered_ser_req = sel_col_ser_req.filter(sel_col_ser_req.Borough == "QUEENS")
```

```python
ct_filtered_ser_req = filtered_ser_req.filter(filtered_ser_req['Complaint Type'] == "Noise - Residential")
```

```python
from pyspark.sql.functions import when
df_with_open = ct_filtered_ser_req.withColumn("is_open", when(ct_filtered_ser_req.Status == "Open", True).otherwise(False))
```
